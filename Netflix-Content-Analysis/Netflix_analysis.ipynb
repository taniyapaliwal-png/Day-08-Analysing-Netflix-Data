# Netflix Data Analysis with Python

## Overview
This project explores a Netflix dataset using Python, Pandas, and Matplotlib.

The analysis focuses on:
- Dataset structure and data quality
- Release-year trends
- User ratings
- Content ratings
- Top-rated titles

```python
## 1. Import the libraries
import pandas as pd 
import matplotlib.pyplot as plt

## 2. Load the data
df = pd.read_csv('Netflix_titles.csv')

## 3. Clean the data
required_cols = ['title', 'rating', 'ratinglevel', 'ratingdescription', 'release_year', 'user_rating_score', 'user_rating_size']
df = df.dropna(subset=[col for col in required_cols if col in df.columns])

## 4. Data visualizations
plt.figure(figsize=(6,4))
rating_counts = df['rating'].value_counts()
plt.bar(rating_counts.index, rating_counts.values, color='skyblue')
plt.title('Netflix Titles by Rating')
plt.xlabel('Rating')
plt.ylabel('Number of Titles')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.savefig('titles_by_rating.png')
plt.show()
```

```python
### 4.1 User Rating Distribution

plt.hist(df["user_rating_score"], bins=10)
plt.title("Distribution of Netflix User Ratings")
plt.xlabel("User Rating Score")
plt.ylabel("Number of Titles")
plt.show()
```

```python
year_counts = df["release_year"].value_counts().sort_index()
```

```python
### 4.2 Netflix Titles by Release Year

plt.bar(year_counts.index, year_counts.values)
plt.title("Number of Netflix Titles by Release Year")
plt.xlabel("Release Year")
plt.ylabel("Number of Titles")
plt.show()
```

```python
### 4.3 Top 10 Netflix Titles by User Rating

plt.barh(top_rated["title"], top_rated["user_rating_score"])

plt.title("Top 10 Netflix Titles by User Rating")
plt.xlabel("User Rating Score")
plt.ylabel("Title")
plt.show()
```

```python
year_counts = df["release_year"].value_counts().sort_index()
```

```python
### 4.5 Netflix Titles Released Over the Years

plt.plot(year_counts.index, year_counts.values)

plt.title("Netflix Titles Released Over the Years")
plt.xlabel("Release Year")
plt.ylabel("Number of Titles")
plt.show()
```
